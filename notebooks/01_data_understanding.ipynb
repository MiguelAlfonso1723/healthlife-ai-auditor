## 4. Identificación de Identificadores

```python
# Definición de llaves basada en la arquitectura de datos (AA-02)
KEYS_MAP = {
    'pacientes': {
        'primary_key': 'id_paciente',
        'foreign_keys': []
    },
    'atenciones': {
        'primary_key': 'id_atencion',
        'foreign_keys': ['id_paciente']
    },
    'historia_clinica': {
        'primary_key': 'id_detalle',
        'foreign_keys': ['id_atencion']
    },
    'prefactura': {
        'primary_key': 'id_prefactura',
        'foreign_keys': ['id_atencion', 'id_paciente']
    },
    'cruce_validacion': {
        'primary_key': 'id_cruce',
        'foreign_keys': ['id_atencion', 'id_prefactura', 'id_detalle_hc']
    }
}

print('Análisis de identificadores por dataset:\n')
for name, keys in KEYS_MAP.items():
    df = datasets[name]
    pk = keys['primary_key']
    pk_unique = df[pk].nunique()
    pk_total = len(df)
    is_unique = pk_unique == pk_total
    
    print(f'📋 {name.upper()}')
    print(f'   Primary Key: {pk}')
    print(f'   Valores únicos: {pk_unique:,} / {pk_total:,} ({"✅ Única" if is_unique else "⚠️ No única"})')
    
    if keys['foreign_keys']:
        print(f'   Foreign Keys: {", ".join(keys["foreign_keys"])}')
        for fk in keys['foreign_keys']:
            if fk in df.columns:
                fk_unique = df[fk].nunique()
                fk_nulls = df[fk].isnull().sum()
                print(f'     - {fk}: {fk_unique:,} valores únicos, {fk_nulls} nulos')
            else:
                print(f'     - {fk}: ⚠️ COLUMNA NO ENCONTRADA')
    print()
```

## 5. Calidad Básica de Datos

```python
def analyze_data_quality(df, name):
    """Genera un reporte de calidad básica para un DataFrame."""
    print(f'\n{"=" * 60}')
    print(f'📊 CALIDAD DE DATOS — {name.upper()}')
    print(f'{"=" * 60}')
    
    # Valores nulos
    nulls = df.isnull().sum()
    null_pct = (df.isnull().sum() / len(df) * 100).round(2)
    
    print(f'\n--- Valores Nulos ---')
    null_report = pd.DataFrame({
        'nulos': nulls,
        'porcentaje': null_pct
    })
    print(null_report[null_report['nulos'] > 0].to_string() if null_report['nulos'].sum() > 0 else '  Sin valores nulos ✅')
    
    # Valores únicos
    print(f'\n--- Valores Únicos ---')
    unique_report = pd.DataFrame({
        'unicos': df.nunique(),
        'total': len(df),
        'pct_unico': (df.nunique() / len(df) * 100).round(2)
    })
    print(unique_report.to_string())
    
    # Duplicados
    duplicates = df.duplicated().sum()
    print(f'\n--- Duplicados ---')
    print(f'  Filas duplicadas: {duplicates:,} ({duplicates / len(df) * 100:.2f}%)')
    
    # Memoria
    memory_mb = df.memory_usage(deep=True).sum() / 1024 / 1024
    print(f'\n--- Memoria ---')
    print(f'  Uso de memoria: {memory_mb:.2f} MB')
    
    return null_report
```

```python
quality_reports = {}
for name, df in datasets.items():
    quality_reports[name] = analyze_data_quality(df, name)
```

## 6. Resumen Estructural

```python
# Descripciones de cada dataset según la arquitectura de datos
DESCRIPTIONS = {
    'pacientes': 'Información demográfica y administrativa de los pacientes.',
    'atenciones': 'Registro de cada atención médica realizada al paciente.',
    'historia_clinica': 'Detalle de procedimientos, tratamientos y exámenes registrados durante la atención.',
    'prefactura': 'Servicios incluidos en la pre-factura antes de su emisión definitiva.',
    'cruce_validacion': 'Resultado histórico del cruce de validación entre Historia Clínica y Pre-factura.'
}

# Generar tabla resumen
summary_rows = []
for name, df in datasets.items():
    keys = KEYS_MAP[name]
    summary_rows.append({
        'Dataset': name,
        'Archivo': DATASET_FILES[name],
        'Registros': f'{df.shape[0]:,}',
        'Columnas': df.shape[1],
        'Primary Key': keys['primary_key'],
        'Foreign Keys': ', '.join(keys['foreign_keys']) if keys['foreign_keys'] else '—',
        'Descripción': DESCRIPTIONS[name]
    })

summary_df = pd.DataFrame(summary_rows)
summary_df
```

```python
# Resumen de calidad consolidado
quality_summary_rows = []
for name, df in datasets.items():
    total_nulls = df.isnull().sum().sum()
    total_cells = df.shape[0] * df.shape[1]
    duplicates = df.duplicated().sum()
    memory_mb = df.memory_usage(deep=True).sum() / 1024 / 1024
    
    quality_summary_rows.append({
        'Dataset': name,
        'Registros': f'{df.shape[0]:,}',
        'Nulos Totales': f'{total_nulls:,}',
        'Pct Nulos': f'{total_nulls / total_cells * 100:.2f}%',
        'Duplicados': f'{duplicates:,}',
        'Memoria (MB)': f'{memory_mb:.2f}'
    })

quality_summary_df = pd.DataFrame(quality_summary_rows)
quality_summary_df
```

## 7. Observaciones

### 7.1 Pacientes (`01_pacientes.csv`)

- **Representa:** Información demográfica y administrativa de los pacientes.
- **Aporta al sistema:** Datos contextuales del paciente (edad, sexo, EPS, tipo de afiliación, ciudad) que permiten segmentar análisis y evaluar posibles sesgos del modelo.
- **Relación:** Es la entidad raíz. Se conecta con `atenciones` y `prefactura` mediante `id_paciente`.

### 7.2 Atenciones (`02_atenciones.csv`)

- **Representa:** Cada evento de servicio de salud prestado a un paciente.
- **Aporta al sistema:** El diagnóstico principal (CIE-10), tipo de atención, fecha y sede. Estos datos son fundamentales para validar la coherencia entre diagnóstico y procedimientos.
- **Relación:** Conecta pacientes con sus registros clínicos y de facturación. Es la tabla pivote del modelo de datos (FK: `id_paciente`; referenciada por: `historia_clinica`, `prefactura`, `cruce_validacion`).

### 7.3 Historia Clínica Detalle (`03_historia_clinica_detalle.csv`)

- **Representa:** El detalle de procedimientos, consultas, tratamientos y exámenes registrados durante cada atención.
- **Aporta al sistema:** Es el soporte clínico que debe justificar cada cobro en la pre-factura. Contiene `codigo_cups`, `tipo_item`, `cantidad_realizada` y `soporte_clinico`.
- **Relación:** Se vincula a `atenciones` mediante `id_atencion`. Cada atención puede tener múltiples registros clínicos.

### 7.4 Pre-factura (`04_prefactura.csv`)

- **Representa:** Los servicios y procedimientos que serán cobrados antes de emitir la factura definitiva.
- **Aporta al sistema:** Contiene `codigo_cups_facturado`, `cantidad_facturada`, `valor_unitario` y `valor_total`. Estos campos se comparan contra la Historia Clínica para detectar inconsistencias.
- **Relación:** Se vincula a `atenciones` y `pacientes` mediante `id_atencion` e `id_paciente`.

### 7.5 Cruce de Validación (`05_cruce_validacion.csv`)

- **Representa:** El resultado histórico de la validación entre Historia Clínica y Pre-factura.
- **Aporta al sistema:** Funciona como ground truth para entrenamiento y evaluación del modelo de IA. Contiene `resultado`, `tipo_alerta` y `severidad`.
- **Relación:** Conecta `atenciones`, `prefactura` e `historia_clinica` mediante `id_atencion`, `id_prefactura` e `id_detalle_hc`.
